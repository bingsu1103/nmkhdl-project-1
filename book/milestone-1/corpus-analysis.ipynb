# Corpus Analysis — Dung lượng source và references

Phân tích dưới đây sử dụng số liệu tổng hợp của shard Member 5. Các phép tính trung bình trên version sử dụng 7.806 version tasks làm mẫu số; chúng không phải trung bình riêng của các archive tải thành công.


```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

version_tasks = 7806
archive_bytes = 30_901_471_478
filtered_tex_bib_bytes = 1_645_848_429

source_sizes = pd.DataFrame([
    {"dataset": "Source archives", "total_bytes": archive_bytes},
    {"dataset": "Extracted TeX/Bib", "total_bytes": filtered_tex_bib_bytes},
])
source_sizes["average_bytes_per_version_task"] = source_sizes["total_bytes"] / version_tasks
source_sizes["total_gib"] = source_sizes["total_bytes"] / (1024 ** 3)
display(source_sizes)
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(source_sizes["dataset"], source_sizes["total_gib"])
ax.set_title("Total source size before and after extraction (Member 5)")
ax.set_ylabel("Total size (GiB)")
ax.set_xlabel("Data stage")
ax.tick_params(axis="x", rotation=10)
fig.tight_layout()
plt.show()
```

```python
references = pd.DataFrame([
    {"category": "Raw references", "count": 198107},
    {"category": "Without arXiv ID", "count": 105544},
    {"category": "Eligible final references", "count": 92563},
])
display(references)
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(references["category"], references["count"])
ax.set_title("Reference coverage (Member 5)")
ax.set_ylabel("Number of references")
ax.set_xlabel("Reference category")
ax.tick_params(axis="x", rotation=15)
fig.tight_layout()
plt.show()
```
