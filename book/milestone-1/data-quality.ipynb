# Data Quality — Chất lượng và tính đầy đủ

Các bảng dưới đây được tạo từ số liệu tổng hợp đã xác nhận của shard Member 5 (5.000 paper IDs). Chúng không đại diện cho toàn bộ nhóm.


```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

discovery = pd.DataFrame([
    {"metric": "Assigned paper IDs", "count": 5000},
    {"metric": "Discovery success", "count": 5000},
    {"metric": "Discovery failure", "count": 0},
])
source = pd.DataFrame([
    {"status": "success", "count": 7217},
    {"status": "no_tex_bib", "count": 543},
    {"status": "http_error", "count": 41},
    {"status": "invalid_archive", "count": 5},
])
s2 = pd.DataFrame([
    {"status": "ok", "count": 4994},
    {"status": "not_found", "count": 6},
    {"status": "failed", "count": 0},
])

display(discovery)
display(source.assign(rate=source["count"] / 7806))
display(s2.assign(rate=s2["count"] / 5000))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
axes[0].bar(["Success", "Failure"], [5000, 0])
axes[0].set_title("Discovery (n=5,000)")
axes[0].set_ylabel("Papers")

axes[1].bar(source["status"], source["count"])
axes[1].set_title("Source status (n=7,806)")
axes[1].set_ylabel("Version tasks")
axes[1].tick_params(axis="x", rotation=35)

axes[2].bar(s2["status"], s2["count"])
axes[2].set_title("Semantic Scholar (n=5,000)")
axes[2].set_ylabel("Papers")

fig.tight_layout()
plt.show()
```

## Validation

Validation đã trả về `is_valid: true`, không có errors hoặc warnings. Reconciliation: 5.000 assigned, 5.000 discovery success, 0 failed discovery, 0 unaccounted và 0 unexpected. Source completeness: 7.806 expected version tasks, 7.806 status records, không thiếu, không non-terminal, không duplicate và không unexpected.

Cần phân biệt **status completeness** với **download success**: tất cả 7.806 version tasks đã có terminal status, nhưng 41 `http_error` và 5 `invalid_archive` không phải source download thành công.
