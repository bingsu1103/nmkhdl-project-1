# Performance — Hiệu năng và tài nguyên

Đề bài yêu cầu benchmark chính thức trên Google Colab CPU-only. Hiện tại chỉ có thể xác nhận tổng dung lượng archive tải về và tổng dung lượng TeX/Bib được giữ lại. Không có log đo kiểm chứng cho wall-clock runtime, peak/average RAM hoặc peak disk usage; các ô này được để trống thay vì ước lượng.


```python
import pandas as pd
from IPython.display import display

version_tasks = 7806
archive_bytes = 30_901_471_478
filtered_tex_bib_bytes = 1_645_848_429

metrics = pd.DataFrame([
    {"metric": "Total source archive bytes", "value": archive_bytes, "status": "measured aggregate"},
    {"metric": "Total extracted TeX/Bib bytes", "value": filtered_tex_bib_bytes, "status": "measured aggregate"},
    {"metric": "Average archive bytes per version task", "value": archive_bytes / version_tasks, "status": "derived average"},
    {"metric": "Average TeX/Bib bytes per version task", "value": filtered_tex_bib_bytes / version_tasks, "status": "derived average"},
    {"metric": "End-to-end runtime seconds", "value": None, "status": "not measured"},
    {"metric": "Discovery runtime seconds", "value": None, "status": "not measured"},
    {"metric": "Average runtime per requested paper seconds", "value": None, "status": "not measured"},
    {"metric": "Peak RAM bytes", "value": None, "status": "not measured"},
    {"metric": "Average RAM bytes", "value": None, "status": "not measured"},
    {"metric": "Peak disk usage bytes", "value": None, "status": "not measured"},
    {"metric": "Final output size bytes", "value": None, "status": "not measured"},
])
display(metrics)
```

## Kế hoạch benchmark còn thiếu

Cần chạy benchmark end-to-end trên một `data-dir` sạch ở Google Colab CPU-only. Ghi lại wall time toàn pipeline và từng stage; lấy mẫu RAM định kỳ; đo peak disk usage; ghi kích thước output cuối cùng. Cần lưu log, Run ID, commit SHA, môi trường và checksum để kết quả có thể kiểm toán.

Các tổng dung lượng ở trên không phải peak disk usage. Không được dùng chúng để thay thế phép đo peak disk, RAM hoặc thời gian.
